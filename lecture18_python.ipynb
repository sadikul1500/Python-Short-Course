# Regression

```python
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
# Data: Study hours vs Marks
X = np.array([2, 4, 6, 8, 10]).reshape(-1, 1)
y = np.array([50, 65, 75, 85, 95])

model = LinearRegression()
model.fit(X, y)
```

```python
new_hours = np.array([[5]])
predicted_marks = model.predict(new_hours)
print("Predicted marks for 5 study hours:", predicted_marks)
```

```python
new_hours = np.array([[3,7,8]]).reshape(-1, 1)
predicted_marks = model.predict(new_hours)
print("Predicted marks for 5 study hours:", predicted_marks)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
import seaborn as sns
```

```python
df = pd.read_csv("/content/house_prices_linear.csv")

df.head()
```

```python
df.corr()
```

```python
sns.heatmap(df.corr(), annot=True)
```

```python
# Target variable
y = df["Price"]
# Features
x = df[["Size", "Bedrooms", "Age"]]
```

```python
from sklearn.model_selection import train_test_split

# Split data: 80% training, 20% testing
# random_state ensures reproducibility
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=50)
```

```python
regressor = LinearRegression() # Initialize model
regressor.fit(x_train, y_train)
```

```python
print("Intercept:", regressor.intercept_)

# Coefficients: effect of each feature on price
print("Coefficients:", regressor.coef_)
```

```python
y_pred = regressor.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)

r2 = r2_score(y_test, y_pred)
print("R^2 Score:", r2)
```

```python
from sklearn.tree import DecisionTreeRegressor

# Initialize and train a Decision Tree Regressor model
dt_regressor = DecisionTreeRegressor(random_state=50) # Setting random_state for reproducibility
dt_regressor.fit(x_train, y_train)
```

```python
# Make predictions with the Decision Tree model
y_pred_dt = dt_regressor.predict(x_test)

# Evaluate the Decision Tree model
mse_dt = mean_squared_error(y_test, y_pred_dt)
r2_dt = r2_score(y_test, y_pred_dt)

print("Decision Tree Mean Squared Error:", mse_dt)
print("Decision Tree R^2 Score:", r2_dt)
```

```python
import pandas as pd

df = pd.read_csv('/content/house_prices_nonlinear.csv')
df.head()
```

```python
# Target variable
y = df["Price"]
# Features
x = df[["Size", "Bedrooms", "Age"]]
```

```python
from sklearn.model_selection import train_test_split

# Split data: 80% training, 20% testing
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=50)
```

```python
from sklearn.tree import DecisionTreeRegressor

dt_regressor = DecisionTreeRegressor(random_state=50)
dt_regressor.fit(x_train, y_train)
```

```python
y_pred = dt_regressor.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)

r2 = r2_score(y_test, y_pred)
print("R^2 Score:", r2)
```
